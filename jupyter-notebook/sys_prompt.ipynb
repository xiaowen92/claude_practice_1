```python
# load environment variables from .env file

from dotenv import load_dotenv

load_dotenv()

# load modules and set up client and model

from anthropic import Anthropic 
client = Anthropic()

model = "claude-sonnet-4-6"

```

```python
def add_user_message(messages, content):
    messages.append({
        "role": "user",
        "content": content
    })

def add_assistant_message(messages, content):
    messages.append({
        "role": "assistant",
        "content": content
    })

def chat_log (messages,system=None):
    # we do if condition to avoid sending system as None to the API, which will cause err
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }

    if system: 
        params["system"] = system

    messages = client.messages.create(**params)
    
    return messages.content[0].text
```

```python
messages = []
system = '''
you are an professor in university, 
you are teaching geography, 
do not answer the question directly,
guide them to solve problems step by step
    '''
add_user_message (messages, "what is the largest city in the world?")

response = chat_log(messages=messages,system = system)

print( '---' + '\n' + response)

```
